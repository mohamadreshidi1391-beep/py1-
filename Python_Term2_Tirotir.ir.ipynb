<h2>۱. برنامه‌نویسی شیءگرا (Object-Oriented Programming - OOP)</h2>
<p>OOP یک پارادایم برنامه‌نویسی است که بر اساس «شیء»ها سازماندهی می‌شود. هر شیء داده‌ها (ویژگی‌ها) و کدهایی (متدها) را که روی آن داده‌ها عمل می‌کنند، در خود جای داده است. این رویکرد به مدیریت پروژه‌های بزرگ کمک شایانی می‌کند.</p>
<p>مفاهیم اصلی OOP عبارتند از:</p>
<ul>
    <li><b>کلاس (Class):</b> یک قالب یا نقشه برای ساخت شیءها.</li>
    <li><b>شیء (Object):</b> یک نمونه واقعی از یک کلاس.</li>
    <li><b>وراثت (Inheritance):</b> یک کلاس می‌تواند ویژگی‌ها و متدهای کلاس دیگر را به ارث ببرد.</li>
    <li><b>چندریختی (Polymorphism):</b> اشیاء مختلف می‌توانند به یک پیام (فراخوانی یک متد) به روش‌های متفاوتی پاسخ دهند.</li>
</ul>
</div>

<div align="right">
<h3>۱.۱. تعریف کلاس و ساخت شیء</h3>
<p>با استفاده از کلمه کلیدی <code>class</code> یک کلاس تعریف می‌کنیم. تابع <code>__init__</code> (سازنده) برای مقداردهی اولیه به ویژگی‌های یک شیء جدید استفاده می‌شود. <code>self</code> به خود شیء اشاره دارد.</p>
</div>

```python
# تعریف یک کلاس به نام Dog
class Dog:
    # این تابع سازنده (Constructor) کلاس است
    def __init__(self, name, age):
        # ویژگی‌های (Attributes) شیء
        self.name = name
        self.age = age
        print(f"نمونه‌ای جدید از سگ به نام {self.name} ساخته شد.")

    # این یک متد (Method) است
    def bark(self):
        return f"{self.name} می‌گوید: واف!"

# ساخت دو شیء (نمونه) از کلاس Dog
dog1 = Dog("رکس", 5)
dog2 = Dog=("شای", 3)

# دسترسی به ویژگی‌های اشیاء
print(f"نام سگ اول: {dog1.name}, سن: {dog1.age}")
print(f"نام سگ دوم: {dog2.name}, سن: {dog2.age}")

# فراخوانی متدهای اشیاء
print(dog1.bark())
print(dog2.bark())
```

<h3>۱.۲. وراثت (Inheritance)</h3>
<p>وراثت به ما اجازه می‌دهد تا یک کلاس جدید (کلاس فرزند) تعریف کنیم که ویژگی‌ها و متدهای یک کلاس موجود (کلاس والد) را به ارث می‌برد. این کار از تکرار کد جلوگیری می‌کند.</p>
</div>

```python
# کلاس والد
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError "کلاس فرزند باید این متد را پیاده‌سازی کند."

# کلاس فرزند که از Animal ارث‌بری می‌کند
class Cat(Animal):
    def speak(self):
        return f"{self.name} می‌گوید: میو!"

# کلاس فرزند دیگر
class Bird(Animal):
    def speak(self):
        return f"{self.name} می‌گوید: جیغ جیغ!"

# ساخت اشیاء از کلاس‌های فرزند
my_cat = Cat="پوکی")
my_bird = Bird="سیمرغ")

print(my_cat.speak())
print(my_bird.speak())
```

<h2>۲. کار با فایل‌ها (File I/O)</h2>
<p>پایتون امکان خواندن از فایل‌ها و نوشتن در آن‌ها را به سادگی فراهم می‌کند. بهترین روش برای کار با فایل‌ها استفاده از بلوک <code>with</code> است که به طور خودکار فایل را می‌بندد.</p>
</div>

```python
# نوشتن در یک فایل (mode='w' - write)
# اگر فایل وجود داشته باشد، محتوای آن پاک می‌شود.
file_path = 'data.txt'
with open(file_path, 'w', encoding='utf-8') as f:
    f.write("سلام، این یک خط متن است.\n")
    f.write("این خط دوم است.\n")

print(f"متن با موفقیت در فایل {file_path} نوشته شد.")

# خواندن از فایل (mode='r' - read)
with open(file_path, 'r', encoding='utf-8') as f:
    content = f.read() # کل محتوای فایل را می‌خواند

print("--- محتوای فایل ---")
print(content)
print("--------------------")

# اضافه کردن به انتهای فایل (mode='a' - append)
with open(file_path, 'a', encoding='utf-8') as f:
    f.write("این خط به انتهای فایل اضافه شده است.\n")

print("متن جدید به انتهای فایل اضافه شد.")

# خواندن خط به خط از فایل
with open(file_path, 'r', encoding='utf-8') as f:
    print("--- خواندن خط به خط ---")
    for line in f:
        print(line.strip()) # .strip() فضاهای خالی و کاراکتر خط جدید را حذف می‌کند
```

<h3>۳.۱. آرگومان‌های متغیر (*args و **kwargs)</h3>
<p>گاهی اوقات نمی‌دانیم چند آرگومان به یک تابع پاسخ داده خواهد شد. در این موارد از <code>*args</code> برای آرگومان‌های موقعیتی و <code>**kwargs</code> برای آرگومان‌های کلیدواژه‌ای استفاده می‌کنیم.</p>
</div>

```python
# تابعی با *args برای دریافت تعداد نامعلومی آرگومان
def sum_all(*numbers):
    """تمام اعداد دریافتی را جمع می‌کند"""
    total = 0
    for num in numbers:
        total += num
    return total

print(f"مجموع اعداد: {sum_all(1, 2, 3, 4, 5)}")
print(f"مجموع اعداد: {sum_all(10, 20)}")

print("-" * 20)

# تابعی با **kwargs برای دریافت تعداد نامعلومی آرگومان کلیدواژه‌ای
def print_info(**info):
    """اطلاعات کلید-مقدار را چاپ می‌کند"""
    for key, value in info.items():
        print(f"{key}: {value}")

print_info(name="علی", age=30, city="تهران")
```

<h3>۳.۲. دکوراتورها (Decorators)</h3>
<p>دکوراتورها توابعی هستند که رفتار یا عملکرد یک تابع دیگر را بدون تغییر کد اصلی آن، تغییر می‌دهند. آن‌ها برای افزودن قابلیت‌هایی مانند لاگ‌گیری، تست عملکرد، احراز هویت و... بسیار مفید هستند.</p>
</div>

```python
# یک دکوراتور ساده برای زمان‌سنجی اجرای یک تابع
import time

def timer_decorator(func):
    def wrapper():
        start_time = time.time()
        func() # فراخوانی تابع اصلی
        end_time = time.time()
        print(f"تابع {func.__name__} در {end_time - start_time:.4f} ثانیه اجرا شد.")
    return wrapper

@timer_decorator # استفاده از دکوراتور
def slow_function():
    """یک تابع که شبیه‌سازی یک عملیات زمان‌بر است"""
    print("شروع عملیات زمان‌بر...")
    time.sleep(2) # 2 ثانیه صبر می‌کند
    print("عملیات تمام شد.")

# فراخوانی تابع
slow_function()
```

<h2>۴. ژنراتورها (Generators)</h2>
<p>ژنراتورها راهی برای ایجاد توالی‌ها هستند. برخلاف توابع عادی که یک مقدار را با <code>return</code> برمی‌گردانند و تمام می‌شوند، ژنراتورها با استفاده از کلمه کلیدی <code>yield</code> مقادیر را یکی پس از دیگری تولید می‌کنند. این کار باعث صرفه‌جویی در حافظه می‌شود، زیرا کل توالی در حافظه بارگذاری نمی‌شود.</p>
</div>

```python
# یک تابع معمولی که لیستی از اعداد را برمی‌گرداند
def create_list(n):
    numbers = []
    for i in range(n):
        numbers.append(i)
    return numbers

# یک ژنراتور که اعداد را یکی یکی تولید می‌کند
def create_generator(n):
    for i in range(n):
        yield i # به جای return از yield استفاده می‌کنیم

# استفاده از تابع معمولی
my_list = create_list(5)
print(f"لیست تولید شده: {my_list}")
print(f"نوع آن: {type(my_list)}")

print("-" * 20)

# استفاده از ژنراتور
my_generator = create_generator(5)
print(f"ژنراتور تولید شده: {my_generator}")
print(f"نوع آن: {type(my_generator)}")

# برای دسترسی به مقادیر ژنراتور، آن را در یک حلقه قرار می‌دهیم
print("مقادیر تولید شده توسط ژنراتور:")
for number in my_generator:
    print(number)
```

<h2>۵. کار با ماژول‌های استاندارد (مثال: datetime)</h2>
<p>پایتون کتابخانه استاندارد غنی دارد. ماژول <code>datetime</code> برای کار با تاریخ و زمان بسیار کاربردی است.</p>
</div>

```python
from datetime import datetime, timedelta

# دریافت تاریخ و زمان فعلی
now = datetime.now()
print(f"تاریخ و زمان فعلی: {now}")

# فرمت‌بندی تاریخ و زمان
formatted_now = now.strftime("%Y/%m/%d %H:%M:%S")
print(f"تاریخ فرمت‌بندی شده: {formatted_now}")

# کار با تاریخ‌های مشخص
new_year = datetime(2024, 1, 1)
print(f"سال نو: {new_year}")

# محاسبه فاصله زمانی
time_until_new_year = new_year - now
print(f"تا سال نو {time_until_new_year.days} روز باقی مانده است.")

# اضافه کردن یک بازه زمانی
next_week = now + timedelta(weeks=1)
print(f"تاریخ هفته آینده: {next_week.strftime('%Y-%m-%d')}")
```
