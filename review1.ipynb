```python
# 4. یک تابع ساده:
def greet(name):
  """این تابع یک پیام خوشامدگویی چاپ می‌کند."""
  print("سلام " + name + "!")

greet("علی")
greet("مریم")
```

```python
# 6. حلقه `while` تا زمانی که یک شرط برقرار است:
count = 0
while count < 3:
  print("شمارنده:", count)
  count += 1
```

```python

# 10. محاسبه فاکتوریل:
def factorial(n):
  """این تابع فاکتوریل یک عدد را محاسبه می‌کند."""
  if n == 0:
    return 1
  else:
    return n * factorial(n-1)

result = factorial(5)
print("فاکتوریل 5 برابر است با:", result)
```
