<h2>۲. متغیرها و انواع داده (Variables and Data Types)</h2>
<p>در پایتون، برای ذخیره داده‌ها از «متغیر»ها استفاده می‌کنیم. نیازی به تعریف نوع متغیر از قبل نیست و پایتون به صورت خودکار نوع آن را تشخیص می‌دهد.</p>
<p><b>انواع داده اصلی:</b></p>
<ul>
    <li><code>int</code>: اعداد صحیح (مثال: ۱۰, -۵, ۰)</li>
    <li><code>float</code>: اعداد اعشاری (مثال: ۳.۱۴, -۲.۵)</li>
    <li><code>str</code>: رشته یا متن (مثال: "سلام پایتون", 'A')</li>
    <li><code>bool</code>: مقادیر منطقی (مقادیر ممکن: <code>True</code> یا <code>False</code>)</li>
</ul>
</div>

```python
# تعریف متغیرها از انواع مختلف
name = "سارا"  # متغیری از نوع رشته (str)
age = 25        # متغیری از نوع عدد صحیح (int)
gpa = 3.8       # متغیری از نوع عدد اعشاری (float)
is_enrolled = True # متغیری از نوع بولین (bool)

# چاپ متغیرها و نوع آن‌ها با استفاده از تابع type()
print(f"نام: {name}، نوع: {type(name)}")
print(f"سن: {age}، نوع: {type(age)}")
print(f"معدل: {gpa}، نوع: {type(gpa)}")
print(f"ثبت‌نام کرده؟: {is_enrolled}، نوع: {type(is_enrolled)}")
```

<div align="right">
<h2>۳. عملگرها (Operators)</h2>
<p>از عملگرها برای انجام عملیات روی متغیرها استفاده می‌شود.</p>
<ul>
    <li><b>عملگرهای حسابی:</b> <code>+</code>, <code>-</code>, <code>*</code>, <code>/</code>, <code>//</code> (تقسیم صحیح), <code>%</code> (باقیمانده), <code>**</code> (توان)</li>
    <li><b>عملگرهای مقایسه‌ای:</b> <code>==</code>, <code>!=</code>, <code>&gt;</code>, <code>&lt;</code>, <code>&gt;=</code>, <code>&lt;=</code></li>
    <li><b>عملگرهای منطقی:</b> <code>and</code>, <code>or</code>, <code>not</code></li>
</ul>
</div>

```python
x = 10
y = 3

# عملگرهای حسابی
print(f"جمع: {x + y}")
print(f"تفریق: {x - y}")
print(f"ضرب: {x * y}")
print(f"تقسیم اعشاری: {x / y}")
print(f"تقسیم صحیح: {x // y}")
print(f"باقیمانده: {x % y}")
print(f"توان: {x ** y}")

print("-" * 20)

# عملگرهای مقایسه‌ای (نتیجه همیشه True یا False است)
print(f"آیا x برابر y است؟ {x == y}")
print(f"آیا x بزرگتر از y است؟ {x > y}")

print("-" * 20)

# عملگرهای منطقی
a = False
b = True
print(f"a and b: {a and b}")
print(f"a or b: {a or b}")
print(f"not a: {not a}")
```

<p>پایتون ساختارهای داده داخلی قدرتمندی برای ذخیره مجموعه‌ای از آیتم‌ها ارائه می‌دهد.</p>

<div align="right">
<h3>۴.۱. لیست‌ها (Lists)</h3>
<p>لیست‌ها مجموعه‌هایی <b>مرتب</b> و <b>قابل تغییر</b> هستند. با استفاده از براکت <code>[]</code> تعریف می‌شوند.</p>
</div>

```python
# تعریف یک لیست
fruits = ["سیب", "پرتقال", "موز", "انگور"]
print(f"لیست میوه‌ها: {fruits}")

# دسترسی به عناصر با ایندکس (ایندکس از ۰ شروع می‌شود)
print(f"اولین میوه: {fruits[0]}")
print(f"آخرین میوه: {fruits[-1]}")

# برش (Slicing)
print(f"دو میوه وسط: {fruits[1:3]}")

# تغییر یک عنصر
fruits[0] = "هلو"
print(f"لیست بعد از تغییر: {fruits}")

# اضافه کردن عنصر جدید با append()
fruits.append("طالبی")
print(f"لیست بعد از اضافه کردن: {fruits}")

# حذف عنصر با remove()
fruits.remove("موز")
print(f"لیست بعد از حذف: {fruits}")
```

<div align="right">
<h3>۴.۲. تاپل‌ها (Tuples)</h3>
<p>تاپل‌ها مانند لیست‌ها هستند اما <b>غیرقابل تغییر</b> (immutable). با استفاده از پرانتز <code>()</code> تعریف می‌شوند. این ویژگی آن‌ها را برای داده‌هایی که نباید تغییر کنند، مناسب می‌سازد.</p>
</div>

```python
# تعریف یک تاپل
coordinates = (35.6892, 51.3890) # طول و عرض جغرافیایی تهران
print(f"مختصات: {coordinates}")

# دسترسی به عناصر مانند لیست
print(f"عرض جغرافیایی: {coordinates[0]}")

# تلاش برای تغییر یک عنصر باعث خطا می‌شود
try:
    coordinates[0] = 36.0
except TypeError as e:
    print(f"خطا: {e}")

# Unpacking: استخراج مقادیر تاپل در متغیرهای جداگانه
lat, lon = coordinates
print(f"عرض: {lat}, طول: {lon}")
```

<div align="right">
<h3>۴.۳. دیکشنری‌ها (Dictionaries)</h3>
<p>دیکشنری‌ها مجموعه‌هایی <b>غیرمرتب</b>، <b>قابل تغییر</b> و <b>اندیس‌گذاری شده</b> هستند. داده‌ها به صورت جفت <b>کلید: مقدار</b> (key: value) ذخیره می‌شوند. با استفاده از آکولاد <code>{}</code> تعریف می‌شوند.</p>
</div>

```python
# تعریف یک دیکشنری
student = {
    "name": "رضا",
    "age": 22,
    "major": "هوش مصنوعی",
    "courses": ["پایتون", "یادگیری ماشین", "آمار"]
}
print(f"اطلاعات دانشجو: {student}")

# دسترسی به مقدار با استفاده از کلید
print(f"رشته تحصیلی: {student['major']}")

# تغییر یک مقدار
student['age'] = 23
print(f"سن جدید: {student['age']}")

# اضافه کردن یک جفت کلید-مقدار جدید
student['gpa'] = 3.9
print(f"دیکنشری بعد از اضافه کردن معدل: {student}")

# دسترسی به تمام کلیدها و مقادیر
print(f"کلیدها: {student.keys()}")
print(f"مقادیر: {student.values()}")
```

<h3>۴.۴. مجموعه‌ها (Sets)</h3>
<p>مجموعه‌ها مجموعه‌هایی <b>غیرمرتب</b> و <b>بدون تکرار</b> هستند. برای انجام عملیات ریاضی مانند اجتماع و اشتراک بسیار مفیدند. با استفاده از آکولاد <code>{}</code> یا تابع <code>set()</code> تعریف می‌شوند.</p>
</div>

```python
# تعریف یک مجموعه
unique_numbers = {1, 2, 3, 2, 4, 5, 1}
print(f"مجموعه اعداد یکتا: {unique_numbers}") # تکرارها به صورت خودکار حذف می‌شوند

# اضافه کردن عنصر
unique_numbers.add(6)
print(f"بعد از اضافه کردن ۶: {unique_numbers}")

# عملیات مجموعه‌ها
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(f"اجتماع: {set_a.union(set_b)}")
print(f"اشتراک: {set_a.intersection(set_b)}")
print(f"تفاضل (A-B): {set_a.difference(set_b)}")
```

<h3>۵.۱. دستورات شرطی (if, elif, else)</h3>
<p>برای اجرای قطعه کدی فقط در صورتی که یک شرط خاص برقرار باشد، استفاده می‌شود.</p>
</div>

```python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "D"

print(f"نمره: {score}, مقطع: {grade}")
```

<div align="right">
<h3>۵.۲. حلقه‌ها (Loops)</h3>
<p>برای تکرار اجرای یک قطعه کد استفاده می‌شوند.</p>
</div>

```python
# حلقه for برای پیمایش در یک لیست
colors = ["قرمز", "سبز", "آبی"]
print("پیمایش لیست رنگ‌ها با for:")
for color in colors:
    print(f"- {color}")

print("-" * 20)

# حلقه for با تابع range() برای تکرار به تعداد مشخص
print("شمارش معکوس با for و range:")
for i in range(5, 0, -1): # از 5 تا 1 (یک واحد کمتر از 0)
    print(i)
print("پایان!")

print("-" * 20)

# حلقه while تا زمانی که شرط درست باشد ادامه می‌یابد
print("شمارش با while:")
count = 0
while count < 5:
    print(f"شمار: {count}")
    count += 1 # معادل count = count + 1
```

<div align="right">
<h3>۵.۳. دستورات break و continue</h3>
<ul>
    <li><code>break</code>: حلقه را به طور کامل متوقف می‌کند.</li>
    <li><code>continue</code>: به تکرار بعدی حلقه می‌رود و بقیه کدهای همان تکرار را نادیده می‌گیرد.</li>
</ul>
</div>

```python
# مثال break
print("مثال break:")
for i in range(10):
    if i == 5:
        print "رسیدیم به 5، حلقه متوقف شد."
        break
    print(i)

print("-" * 20)

# مثال continue
print("مثال continue:")
for i in range(5):
    if i == 2:
        print "عدد 2 نادیده گرفته شد."
        continue
    print(i)
```

<h2>۶. توابع (Functions)</h2>
<p>توابع بلوک‌های کد قابل استفاده مجدد هستند که یک وظیفه مشخص را انجام می‌دهند. آن‌ها با کلمه کلیدی <code>def</code> تعریف می‌شوند.</p>
</div>

```python
# تعریف یک تابع ساده بدون ورودی و خروجی
def greet():
    """این تابع یک پیام خوشامدگویی چاپ می‌کند."""
    print("سلام! به آموزشگاه هوش مصنوعی tirotir.ir خوش آمدید.")

# فراخوانی تابع
greet()

print("-" * 20)

# تعریف تابع با ورودی (پارامتر) و خروجی (return)
def calculate_area(width, height):
    """مساحت یک مستطیل را محاسبه می‌کند."""
    return width * height

# فراخوانی تابع با ارسال آرگومان
w = 10
h = 5
area = calculate_area(w, h)
print(f"مساحت مستطیل با عرض {w} و ارتفاع {h} برابر است با: {area}")
```

<h3>۷.۱. توابع لامبدا (Lambda Functions)</h3>
<p>توابع لامبدا یا توابع ناشناس، توابع کوچکی هستند که در یک خط تعریف می‌شوند و نیازی به کلمه کلیدی <code>def</code> ندارند. آن‌ها برای عملیات ساده و کوتاه بسیار مناسب هستند.</p>
</div>

```python
# تابع معمولی برای به توان دو رساندن یک عدد
def square_normal(x):
    return x * x

# معادل آن با تابع لامبدا
square_lambda = lambda x: x * x

print(f"نتیجه تابع معمولی برای 7: {square_normal(7)}")
print(f"نتیجه تابع لامبدا برای 7: {square_lambda(7)}")

# استفاده از لامبدا با تابع map برای اعمال یک عمل روی تمام عناصر یک لیست
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x**2, numbers))
print(f"اعداد اولیه: {numbers}")
print(f"مربع اعداد: {squared_numbers}")
```

<div align="right">
<h3>۷.۲. درک لیست (List Comprehension)</h3>
<p>یک روش کوتاه و خوانا برای ساخت لیست‌ها بر اساس یک لیست موجود. این روش جایگزینی مدرن برای حلقه‌های <code>for</code> ساده‌ای است که در آن لیستی ساخته می‌شود.</p>
</div>

```python
# روش سنتی با حلقه for
squares = []
for i in range(10):
    squares.append(i * i)
print(f"روش سنتی: {squares}")

# روش مدرن با List Comprehension
squares_comp = [i * i for i in range(10)]
print(f"روش List Comprehension: {squares_comp}")

# List Comprehension با شرط
even_squares = [i * i for i in range(10) if i % 2 == 0]
print(f"مربع اعداد زوج: {even_squares}")
```

<h3>۷.۳. کار با ماژول‌ها (Modules)</h3>
<p>ماژول‌ها فایل‌های پایتونی هستند که توابع و متغیرهای تعریف شده در آن‌ها را می‌توان در برنامه‌های دیگر استفاده کرد. این کار باعث سازماندهی بهتر و استفاده مجدد کد می‌شود.</p>

```python
# وارد کردن ماژول math برای استفاده از توابع ریاضی
import math

print(f"عدد پی: {math.pi}")
print(f"جذر 16: {math.sqrt(16)}")
print(f"گرد کردن 3.7: {math.ceil(3.7)}")

# وارد کردن تابع مشخص از یک ماژول
from random import randint

# تولید یک عدد صحیح تصادفی بین 1 تا 100
random_num = randint(1, 100)
print(f"یک عدد تصادفی: {random_num}")
```

<h3>۷.۴. مدیریت خطا (Error Handling)</h3>
<p>با استفاده از بلوک <code>try...except</code> می‌توان خطاهای احتمالی را مدیریت کرد و از توقف ناگهانی برنامه جلوگیری نمود.</p>
</div>

```python
try:
    # کدی که ممکن است خطا ایجاد کند
    num_str = "ده"
    num = int(num_str)
    print(f"عدد تبدیل شده: {num}")
except ValueError:
    # این بخش در صورت بروز خطای ValueError اجرا می‌شود
    print(f"خطا: نمی‌توان رشته '{num_str}' را به عدد تبدیل کرد.")
except Exception as e:
    # این بخش در صورت بروز هر خطای دیگری اجرا می‌شود
    print(f"یک خطای پیش‌بینی نشده رخ داد: {e}")
else:
    # این بخش فقط در صورتی که هیچ خطایی رخ ندهد اجرا می‌شود
    print("تبدیل با موفقیت انجام شد.")
finally:
    # این بخش همیشه، چه خطا رخ دهد و چه ندهد، اجرا می‌شود
    print("بلوک مدیریت خطا به پایان رسید.")
```
