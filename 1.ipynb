```python

import random

for i in range(5):
    a = random.randint(1, 6)
    if a == 6:
        print('wow', a)
    else:
        print(a)
```

```python

import random

while True:
    for i in range(5):
        a = random.randint(1, 6)
        if a == 6:
            print('wow', a)
        else:
            print(a)
```

```python
import random

while True:
    a = random.randint(1, 6)
    if a == 6:
        print("wow", a)
        break   # stop the loop
    else:
        print(a)
```

```python

import random

try:
    while True:
        print("Ctrl + C for exit")
        for i in range(5):
            a = random.randint(1, 6)
            if a == 6:
                print('wow', a)
            else:
                print(a)
except KeyboardInterrupt:
    print("bye")
```

```python
rows = 5
for i in range(1, rows+1):
    print(" " * (rows - i) + "*" * i)
```

```python
rows = 5
for i in range(1, rows+1):
    print(" " * (rows - i) + "*" * (2*i - 1))
```

```python
rows = 5
for i in range(rows, 0, -1):
    print(" " * (rows - i) + "*" * (2*i - 1))
```

```python
rows = 5
# upper part
for i in range(1, rows+1):
    print(" " * (rows - i) + "*" * (2*i - 1))
# lower part
for i in range(rows-1, 0, -1):
    print(" " * (rows - i) + "*" * (2*i - 1))
```
